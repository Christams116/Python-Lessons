# 🔄 The while Loop (Repeat Machine)
In Python, we use while loops when we want to do something multiple times without typing the code over and over again.

✅ How a while Loop Works
A while loop needs three main parts:

A Starting Point: Usually a variable set to a number (like count = 1).

A Condition: The rule that tells the loop when to keep going (like while count <= 5).

A Change: Something that happens inside the loop so it eventually ends (like count = count + 1).

Example:

```python
count = 1

while count <= 3:
    print("Loop number: " + str(count))
    count = count + 1  # This makes count go up by 1 each time
```

## ✅ Property 1: The Infinite Loop ♾️
If you forget to "change" your variable (like adding 1 to your count), the loop will never stop! This is called an Infinite Loop.

## ✅ Property 2: Using while with input()
You can use a while loop to keep asking a user a question until they give you the right answer.

Example:

```python
answer = ""

while answer != "please":
    answer = input("What is the magic word? ")

print("You may enter!")
```
