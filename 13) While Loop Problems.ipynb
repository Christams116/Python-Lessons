# 🎮 Task 3: Game Over Loop
Your Goal: Keep asking the user "Do you want to keep playing? (yes/no)". The loop should keep running as long as they type "yes".

```python
# 1. Starting variable:
choice = "yes"

# 2. Write the while loop:
while choice == "yes":
    choice = input("Do you want to keep playing? (yes/no): ")

print("Game Over! 🎮")
```

# 🏆 Task 5: The Secret Number
Your Goal: Set a secret_number to 7. Use while to keep asking the user to "Guess the number" until they get it right.

```python
secret_number = 7
guess = 0

# 1. Loop while the guess is NOT EQUAL (!=) to the secret_number:
while guess != secret_number:
    guess = int(input("Guess the secret number: "))

print("You got it! 🎉")
```
